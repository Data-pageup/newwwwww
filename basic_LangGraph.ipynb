```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

# Problem 1

```python
class TemperatureState(TypedDict):
    temp_celsius:float
    temp_fahrenheit:float 
    weather_status : str 
    
def convert_temp(state: TemperatureState) -> TemperatureState:
    celsius = state['temp_celsius']
    fahrenheit = (celsius * 9/5 ) + 32
    state['temp_fahrenheit'] = round(fahrenheit,2)
    return state

def temp_des(state: TemperatureState) -> TemperatureState:
    fahrenheit = state["temp_fahrenheit"]

    if fahrenheit < 50:
        state["weather_status"] = "Cold"

    elif 50 <= fahrenheit < 77:
        state["weather_status"] = "Mild"

    elif 77 <= fahrenheit < 95:
        state["weather_status"] = "Hot"

    else:
        state["weather_status"] = "Extreme Heat"

    return state

# define graph
graph = StateGraph(TemperatureState)

#add nodes to the graph
graph.add_node("convert_temp",convert_temp)
graph.add_node("temp_des",temp_des)


# add edges to the graph

graph.add_edge(START,"convert_temp")
graph.add_edge("convert_temp","temp_des")

graph.add_edge("temp_des",END)


#compile the graph

workflow = graph.compile()
initial_state = {"temp_celsius": 25}
workflow.invoke(initial_state)
```

# problem 2

```python
# state 

class StudentState(TypedDict):
    name : str
    math :float
    english :float
    social: float
    avg: float
    res : str


# Creating node functions 

# node1

def calculate_average(State:StudentState) -> StudentState:
    a = State['math'] + State['english'] + State["social"]
    State['avg'] = a / 3
    return State

# node 2

def determine_result(State:StudentState) -> StudentState :
    score_avg = State['avg']
    if score_avg >70 :
        State["res"] = "pass"
    else:
        State['res'] = "fail"

 # creating graph

graph = StateGraph(StudentState)

# creating nodes 

graph.add_node("calculate_average",calculate_average)

graph.add_node("determine_result",determine_result)


# creating edges

graph.add_edge(START,"calculate_average")

graph.add_edge("calculate_average","determine_result")

graph.add_edge("determine_result",END)

wf = graph.compile()

initial_state  = {'name': "Ag", "math": 90,"english":95,"social":80}

wf.invoke(initial_state)
```

# Problem 3

```python
 # You receive a person's information and need to determine whether they are eligible for a loan.

class PersonInfo(TypedDict):
    name : str
    age : int 
    monthly_income : int
    credit_score : int
    loan_amount : int
    income_score : str
    credit : str
    output : str
```

```python
# node 1 function 

def check_income(State: PersonInfo) -> PersonInfo :
    income = State['monthly_income']
    if income > 50000 :
        State['income_score'] = "good"
    else:
        State['income_score'] = "bad"

    return State


# node 2 function 

def check_credit(State: PersonInfo) -> PersonInfo:
    if State['credit_score'] > 750 :
        State["credit"] = "good"
    else:
        State['credit'] = "bad"
    return State

# node 3 function 

def loan_decision(State: PersonInfo) -> PersonInfo:
    if State['credit'] == "good" and State['income_score'] == "good" :
        State["output"] = "Approved & Good for getting role"
    else:
        State["output"] = "Sorry, dude you have to work hard in the rat race"

    return State



# creating graph 

graph = StateGraph(PersonInfo)

# creating nodes 

graph.add_node("check_income",check_income)
graph.add_node("check_credit",check_credit)
graph.add_node("loan_decision",loan_decision)


# creatin edges 

graph.add_edge(START,"check_income")
graph.add_edge("check_income","check_credit")
graph.add_edge("check_credit","loan_decision")
graph.add_edge("loan_decision",END)



wf1 = graph.compile()

initial_state = {"name": "Ganbesh","age":24,"monthly_income": 5000,"credit_score": 770,"loan_amount":500000}

wf1.invoke(initial_state)
```
